```python
import folium
from folium.plugins import MarkerCluster, MousePosition
from folium.plugins import FastMarkerCluster

# Initialize map centered around US launch sites
spacex_map = folium.Map(location=[28.573, -80.646], zoom_start=5)

# Add launch site coordinate markers
# Example: NASA Kennedy Space Center LC-39A
ksc_coords = [28.573255, -80.646898]
folium.Marker(
    ksc_coords,
    popup='KSC LC-39A',
    icon=folium.Icon(color='red', icon='rocket')
).add_to(spacex_map)

spacex_map.save("spacex_map.html")
print("Folium map script executed and saved.")
```
