```python
import pandas as pd
import sqlite3

# Create SQLite in-memory database connection
conn = sqlite3.connect('spacex.db')
cursor = conn.cursor()

# Load dataset into SQL table
df = pd.read_csv("dataset_part_2.csv")
df.to_sql("SPACEXTBL", conn, if_exists='replace', index=False)

# Sample SQL Query: Display names of unique launch sites
query_sites = "SELECT DISTINCT LaunchSite FROM SPACEXTBL;"
print(pd.read_sql(query_sites, conn))

# Total payload mass shipped by NASA (CRS)
query_payload = "SELECT SUM(PayloadMass) FROM SPACEXTBL WHERE Customer LIKE '%NASA (CRS)%';"
print(pd.read_sql(query_payload, conn))
```
