```python
import pandas as pd
import dash
from dash import html, dcc
from dash.dependencies import Input, Output
import plotly.express as px

spacex_df = pd.read_csv("dataset_part_2.csv")
app = dash.Dash(__name__)

app.layout = html.Div(children=[
    html.H1('SpaceX Launch Records Dashboard', style={'textAlign': 'center'}),
    dcc.Dropdown(
        id='site-dropdown',
        options=[{'label': 'All Sites', 'value': 'ALL'}, {'label': 'KSC LC-39A', 'value': 'KSC LC-39A'}],
        value='ALL',
        placeholder="Select a Launch Site"
    ),
    html.Div(dcc.Graph(id='success-pie-chart'))
])

@app.callback(
    Output('success-pie-chart', 'figure'),
    Input('site-dropdown', 'value')
)
def update_pie(site):
    if site == 'ALL':
        return px.pie(spacex_df, values='class', names='LaunchSite', title='Total Success by Site')
    else:
        filtered = spacex_df[spacex_df['LaunchSite'] == site]
        return px.pie(filtered, names='class', title=f'Success vs Failure for {site}')

if __name__ == '__main__':
    app.run_server(debug=True)
```
