```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("dataset_part_1.csv")

# Identify missing values
print(df.isnull().sum())

# Determine landing outcome target variable (class: 1 for success, 0 for failure)
landing_outcomes = df['Outcome'].value_counts()
print(landing_outcomes)

# Create training landing binary outcome
def landing_success(outcome):
    if 'True' in outcome:
        return 1
    else:
        return 0

df['Class'] = df['Outcome'].apply(landing_success)
df.to_csv("dataset_part_2.csv", index=False)
print("Data wrangling completed and saved to dataset_part_2.csv")
```
