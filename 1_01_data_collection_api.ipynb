```python
import requests
import pandas as pd
import datetime

# 1. Request past launches from SpaceX API v4
api_url = "https://api.spacexdata.com/v4/launches/past"
response = requests.get(api_url)
data = response.json()

# Helper functions for API unpacking
def getBoosterVersion(data):
    booster_list = []
    for booster_id in data['rocket']:
        if booster_id:
            res = requests.get(f"https://api.spacexdata.com/v4/rockets/{booster_id}").json()
            booster_list.append(res.get('name'))
    return booster_list

def getPayloadData(data):
    payload_list = []
    for payload_id in data['payloads']:
        if payload_id:
            res = requests.get(f"https://api.spacexdata.com/v4/payloads/{payload_id}").json()
            payload_list.append(res)
    return payload_list

def getCoreData(data):
    core_list = []
    for core in data['cores']:
        if core:
            core_list.append(core)
    return core_list

def getLaunchSite(data):
    site_list = []
    for site_id in data['launchpad']:
        if site_id:
            res = requests.get(f"https://api.spacexdata.com/v4/launchpads/{site_id}").json()
            site_list.append(res.get('name'))
    return site_list

print("API Data Collection script ready to execute in Colab.")
```
